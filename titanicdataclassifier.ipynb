```python
train_data=pd.read_csv("/kaggle/input/competitions/titanic/train.csv")
test_data=pd.read_csv("/kaggle/input/competitions/titanic/test.csv")
#com_data=pd.read_csv("/kaggle/input/competitions/titanic/gender_submission.csv")
```

```python
# Her iki veri kümesini bir liste içinde döngüye alarak doldurma
for dataset in [train_data, test_data]:
    dataset['Cabin'] = dataset['Cabin'].ffill().bfill()
```

```python
train_data=train_data.drop(columns=["Name","Parch","Ticket","SibSp","Fare","Embarked"],axis=1)
```

```python
test_data=test_data.drop(columns=["Name","Parch","Ticket","SibSp","Fare","Embarked"],axis=1)
```

```python
train_data
```

```python
train_data.info()
```

```python
y=train_data["Survived"]
X = pd.get_dummies(train_data,columns=["Sex", "Cabin"])
X_test = pd.get_dummies(test_data,columns=["Sex", "Cabin"])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
test_data
```

```python
X, X_test = X.align(X_test, join='left', axis=1, fill_value=0)
```

```python
# 5. Modeli tanımla ve eğit (Train)
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X, y)
```

```python
model.score(X,y)
```

```python
# 6. Test verisi üzerinde tahmin yap (Test / Predict)
predictions = model.predict(X_test)
```

```python
output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
print("Model eğitildi, test edildi ve submission.csv başarıyla kaydedildi!")
```
