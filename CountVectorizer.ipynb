```python
#Count Vectorizer
#Her metni sesi arraya çevirmeliyiz.
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer

```

```python
df = pd.DataFrame()
df["cumleler"]=["ali bak","ali ata bak","bak ali bak","ali güzel ata bak","ışık ılık süt iç","ışık süt iç","iç ışık iç"]
cv=CountVectorizer(max_features=4)
x=cv.fit_transform(df["cumleler"]).toarray()
print(x)
#YOĞUN OLAN KELİMELERİ YAKALIYOR VE KATEGOEİZE EDİYOR.
```

```python
cv.get_feature_names_out() #Önemli kelimelere bu şekilde bakabiliyoruz.
```
