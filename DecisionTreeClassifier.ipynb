```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier,export_graphviz
from sklearn.model_selection import train_test_split
import graphviz
```

```python
df=pd.read_csv('heart.csv')
```

```python
df.head(3)
```

```python
y=df['output']
x=df.drop('output',axis=1)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.09,random_state=19)
```

```python
tree=DecisionTreeClassifier()
model=tree.fit(x_train,y_train) #öğrenme
```

```python
model.score(x_test,y_test)
```

```python
x.head(1)
```

```python
tahmin=model.predict([[25,0,1,230,1,0,100,0,2,0,0,0,1]])
```

```python
if tahmin==1:
  print("kalma ihtimali vardır")
else:
  print("kalma ihtimali yoktur")
```
