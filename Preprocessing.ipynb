```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df=pd.read_csv('audi.csv');
```

```python
df.head(3)
```

```python
df=df.drop(columns=["index","href","MileageRank","PriceRank","PPYRank","Score"])
```

```python
#Ağırlığı etkilemeyen sütünları sildik Önemli !!!
```

```python
df.info()
```

```python
df.head(3)
```

```python
df.columns=["yil","kasa","mil","motor","ps","vites","yakit","sahip","fiyat","ppy"]
```

```python
#Özellikleri türkçeleştirdik
```

```python
df.head(3)
```

```python
df["motor"]=df["motor"].str.replace("L","")
```

```python
df["motor"]=pd.to_numeric(df["motor"])
```

```python
df.head(3)
```

```python
df=pd.get_dummies(df,columns=["kasa","yakit","vites"],drop_first=True)
```

```python
#Bazı özellikleri numerik bir hale çevidik.Özelliklerin ilk halini kaldırdık.
#Veri önişleme işlemini tamamladık.
```

```python
y=df['fiyat']
```

```python
x=df.drop(columns=["fiyat"])
```

```python
#x ve y belirledik.
```

```python
df.head(3)
```

```python
l=LinearRegression()
model=l.fit(x,y)
```

```python
model.predict([[2017,30000,1.6,110,1,2600,0,1]])
```

```python
model.score(x,y)
```
