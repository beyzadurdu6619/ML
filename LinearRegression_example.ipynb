```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('student.csv')
```

```python
df.head(5)
```

```python
df=pd.get_dummies(df,columns=['Extracurricular Activities'],drop_first=True)
```

```python
df.columns=['saat','son_not','uyku_saati','pratik_sayisi','not','faaliyet']
```

```python
df.head(3)
```

```python
df['not']=df['not'].astype(int)
df.head(3)
```

```python
df.info()
```

```python
y=df['not']
x=df.drop('not',axis=1)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
df.head(3)
```

```python
l=LinearRegression()
model=l.fit(x_train,y_train)
```

```python
model.predict([[10,60,8,3,1]])
```

```python
#y=a1*x1+a2*x2+b
```

```python
model.coef_ #katsayı a1,a2,a3...
```

```python
model.intercept_ #sabit
```

```python
y_value=(model.coef_[0]*10)+(model.coef_[1]*60)+(model.coef_[2]*8)+(model.coef_[3]*3)+(model.intercept_)
print(y_value)
```

```python
x1=model.score(x_test,y_test)
```

```python
x2=model.score(x_train,y_train)
```

```python
son=x1-x2
print(son)
```
