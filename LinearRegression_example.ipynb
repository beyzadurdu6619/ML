```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('student.csv')
```

```python
df.head(5)
```

```python
df=pd.get_dummies(df,columns=['Extracurricular Activities'],drop_first=True)
```

```python
df.info()
```

```python
df['Performance Index']=df['Performance Index'].astype(int)
df.head(3)
```

```python
df.info()
```

```python
y=df['Performance Index']
x=df.drop('Performance Index',axis=1)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
df.head(3)
```

```python
l=LinearRegression()
model=l.fit(x_train,y_train)
```

```python
model.predict([[10,60,8,3,1]])
```

```python
x1=model.score(x_test,y_test)
```

```python
x2=model.score(x_train,y_train)
```

```python
son=x1-x2
print(son)
```
