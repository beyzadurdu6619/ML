```python
from prophet import Prophet
import yfinance as yf
```

```python
df=yf.download('BTC-USD',"2015-01-01","2026-08-30")
```

```python
df=df[["Close"]]
```

```python
df.head(3)
```

```python
df=df.reset_index()
```

```python
df.head(3)
```

```python
df.columns=["ds","y"]
```

```python
df.head(3)
```

```python
pro=Prophet()
pro.fit(df)
```

```python
future=pro.make_future_dataframe(360) #100 gün sonrayı tahmin edeceksem 100. güne kadarki günleri de tahmin edebilmeliyim.Bu metod bu durumun DataFrameini oluşturur.
```

```python
tahmin=pro.predict(future)
```

```python
pro.plot(tahmin)
```

```python
pro.plot_components(tahmin)
```
