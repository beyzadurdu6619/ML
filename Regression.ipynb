```python
from sklearn.linear_model import LinearRegression
import pandas as pd
```

```python
#y=mx+b
```

```python
df=pd.read_csv("student.csv")
```

```python
df.head(3)
```

```python
y=df[["Marks"]] #prediction
x=df[["number_courses","time_study"]] #feature
```

```python
df.info()
```

```python
l=LinearRegression()
model=l.fit(x,y) #model learns
```

```python
model.predict([[4,10]])
```

```python
df["Marks"].max()
```

```python
model.score(x,y)
```
